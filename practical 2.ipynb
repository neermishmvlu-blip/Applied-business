```python

import pandas as pd
import matplotlib.pyplot as plt

```

```python
df = pd.read_csv("student_marks.csv")
```

```python
print(df.head())
```

```python
marks = df["Maths Marks"]
```

```python
mean = marks.mean()
median = marks.median()
mode = marks.mode()
```

```python
print("Mean:", mean)
print("Median:", median)
print("Mode:", mode.values)
```

```python
print("Minimum :", marks.min())
```

```python
print("Maximum :", marks.max())
```

```python
print("Range :", marks.max() - marks.min())
```

```python
print("Variance :", marks.var())
```

```python
print("Standard Deviation :", marks.std())
```

```python
print("Count :", marks.count())
```

```python
print("Sum :", marks.sum())
```

```python
print("\nDescriptive Statistics")
print(marks.describe())
```

```python
import seaborn as sns
```

```python
plt.figure(figsize=(8,6))
sns.histplot(marks, bins=5, kde=True, color="skyblue", edgecolor="black")
plt.title("Histogram of Maths Marks")
plt.xlabel("Maths Marks")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8,3))
sns.boxplot(x=marks, color="orange")
plt.title("Box Plot of Maths Marks")
plt.xlabel("Maths Marks")
plt.show()
```

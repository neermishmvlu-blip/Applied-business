```python
import pandas as pd
import matplotlib.pyplot as plt

```

```python
df = pd.read_csv("sales_data.csv")
```

```python
print(df.head())
```

```python
sales = df["Sales"]
```

```python
print("\n----- Measures of Central Tendency -----")
print("Mean :", sales.mean())
print("Median :", sales.median())
print("Mode :")
print(sales)
```

```python
print("\n----- Measures of Dispersion -----")
print("Minimum :", sales.min())
print("Maximum :", sales.max())
print("Range :", sales.max() - sales.min())
print("Variance :", sales.var())
print("Standard Deviation :", sales.std())
```

```python
print("Count :", sales.count())
print("Sum :", sales.sum())
```

```python
print(sales.describe())
```

PRATICAL 2B

```python
import seaborn as sns
```

```python
plt.figure(figsize=(8,6))
sns.histplot(sales, bins=10, kde=True, color="skyblue", edgecolor="black")
plt.title("Histogram of Sales")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8,3))
sns.boxplot(x=sales, color="orange")
plt.title("Box Plot of Sales")
plt.xlabel("Sales")
plt.show()
```
